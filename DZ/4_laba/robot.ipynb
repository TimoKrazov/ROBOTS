```python
import sqlite3
```

```python
def connect_to_db(db='company_data.db'):
    return sqlite3.connect(db)
```

```python
def robot(conn):
    cursor = conn.cursor()
    cursor.execute('''
        CREATE TABLE IF NOT EXISTS project_finance_status (
            project_id INTEGER PRIMARY KEY,
            utilization_percent REAL NOT NULL,
            status TEXT NOT NULL
        )
    ''')

    conn.commit()
    cursor.execute('SELECT project_id, budget, spent FROM projects')
    rows = cursor.fetchall()
    
    result = []

    for project_id, budget, spent in rows:
        utilization = 0.0 if budget == 0 else (spent / budget) * 100
        status = "Рисковые" if utilization > 90 else "Стабильные"
        result.append((project_id, round(utilization,2), status))

    cursor.execute('DELETE FROM project_finance_status')
    cursor.executemany('INSERT INTO project_finance_status (project_id, utilization_percent, status) VALUES (?, ?, ?)', result)
    conn.commit()
    conn.close()
    
```

```python
robot(connect_to_db())
```

```python
conn = connect_to_db()
cursor = conn.cursor()
cursor.execute("SELECT * FROM project_finance_status LIMIT 10;")
rows = cursor.fetchall()
for row in rows:
    print(row)
```
